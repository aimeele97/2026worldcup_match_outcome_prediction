```python
# standardise team name between all data before merging
result_df['away_team'] = result_df['away_team'].replace('São Tomé and Príncipe', 'Sao Tome and Principe')
result_df['home_team'] = result_df['home_team'].replace('São Tomé and Príncipe', 'Sao Tome and Principe')

name_map = {
    'Korea Republic': 'South Korea',
    'USA': 'United States',
    "Côte d'Ivoire": 'Ivory Coast',
    'IR Iran': 'Iran',
    'Czechia': 'Czech Republic',
    'Cabo Verde': 'Cape Verde',
    'Congo DR': 'DR Congo',
    'Curacao': 'Curaçao',
    'Korea DPR': 'North Korea',
    'China PR': 'China',
    'Kyrgyz Republic': 'Kyrgyzstan',
    'Türkiye': 'Turkey',
    'St Lucia': 'Saint Lucia',
    'The Gambia': 'Gambia',
    'St Kitts and Nevis': 'Saint Kitts and Nevis',
    'Hong Kong, China': 'Hong Kong',
    'São Tomé and Príncipe': 'Sao Tome and Principe'
}

rank_2024_df['country_full'] = rank_2024_df['country_full'].replace(name_map)
rank_2026_df['team'] = rank_2026_df['team'].replace(name_map)
```

```python
date_range = pd.date_range(rank_2024_df['rank_date'].min(), rank_2024_df['rank_date'].max(), freq='D')

countries = rank_2024_df['country_full'].unique()

full_grid = pd.MultiIndex.from_product([countries, date_range], names=['country_full', 'rank_date'])
full_grid_df = pd.DataFrame(index=full_grid).reset_index()
new_rank_df = pd.merge(full_grid_df, rank_2024_df, on=['country_full', 'rank_date'], how='left')

# forward fill per country — carries previous rank forward
new_rank_df = new_rank_df.sort_values(['country_full', 'rank_date'])
new_rank_df = new_rank_df.set_index('rank_date')
new_rank_df = new_rank_df.groupby('country_full')[new_rank_df.columns].ffill()
new_rank_df = new_rank_df.reset_index()

print(new_rank_df.columns.tolist())
new_rank_df.head()
```

```python
temp_df = merged_df[merged_df['date'] > pd.to_datetime('2024-06-20')]

# create a lookup dict from your ranking df
rank_lookup = rank_2026_df.set_index('country_full')['rank'].to_dict()
previous_rank_lookup = rank_2026_df.set_index('country_full')['previous_rank'].to_dict()
points_lookup = rank_2026_df.set_index('country_full')['points'].to_dict()
previous_points_lookup = rank_2026_df.set_index('country_full')['previous_points'].to_dict()
rank_change_lookup = rank_2026_df.set_index('country_full')['rank_change'].to_dict()
confederation_lookup = rank_2026_df.set_index('country_full')['confederation'].to_dict()

# fill NaN by mapping country name
temp_df['home_rank'] = temp_df['home_rank'].fillna(temp_df['home_team'].map(rank_lookup))
temp_df['away_rank'] = temp_df['away_rank'].fillna(temp_df['away_team'].map(rank_lookup))
temp_df['home_total_points'] = temp_df['home_total_points'].fillna(temp_df['home_team'].map(points_lookup))
temp_df['away_total_points'] = temp_df['away_total_points'].fillna(temp_df['away_team'].map(points_lookup))
temp_df['home_previous_points'] = temp_df['home_previous_points'].fillna(temp_df['home_team'].map(previous_points_lookup))
temp_df['away_previous_points'] = temp_df['away_previous_points'].fillna(temp_df['away_team'].map(previous_points_lookup))
temp_df['home_rank_change'] = temp_df['home_rank_change'].fillna(temp_df['home_team'].map(rank_change_lookup))
temp_df['away_rank_change'] = temp_df['away_rank_change'].fillna(temp_df['away_team'].map(rank_change_lookup))
temp_df['home_confederation'] = temp_df['home_confederation'].fillna(temp_df['home_team'].map(confederation_lookup))
temp_df['away_confederation'] = temp_df['away_confederation'].fillna(temp_df['away_team'].map(confederation_lookup))
```

#### Feature engineering

```python
def assign_result(home_score, away_score):
    if home_score > away_score:
        return pd.Series([0, 3, 0])
    if home_score < away_score:
        return pd.Series([1, 0, 3])
    else:
        return pd.Series([2, 1, 1])

results = final_df.apply(lambda x: assign_result(x["home_score"], x["away_score"]), axis=1)
final_df[['result', 'home_team_points', 'away_team_points']] = results
```

```python
final_df["rank_dif"] = final_df["home_rank"] - final_df["away_rank"]
final_df["score_dif"] = final_df["home_score"] - final_df["away_score"]
final_df["points_home_by_rank"] = final_df["home_team_points"]/final_df["away_rank"]
final_df["points_away_by_rank"] = final_df["away_team_points"]/final_df["home_rank"]
```

```python
home_df = final_df[['date','home_team', 'home_score', 'away_score', 'home_rank', 'away_rank', 'home_total_points',
       'home_previous_points', 'home_rank_change', 'home_confederation',
       'home_team_points', 'points_home_by_rank', 'result', 'rank_dif']]
home_df.columns = [h.replace("home_", "").replace("_home", "").replace("away_", "suf_").replace("_away", "_suf") for h in home_df.columns]
away_df = final_df[['date','away_team', 'away_score', 'home_score', 'away_rank', 'home_rank', 'away_total_points',
       'away_previous_points', 'away_rank_change', 'away_confederation',
       'away_team_points', 'points_away_by_rank', 'result', 'rank_dif']]
away_df.columns = [h.replace("away_", "").replace("_away", "").replace("home_", "suf_").replace("_home", "_suf") for h in away_df.columns]
```

```python
team_df = pd.concat([home_df, away_df], ignore_index=True)
```

```python
new_val = []
for idx, row in team_df.iterrows():
   team= row['team']
   date = row['date']
   past_games = team_df.loc[(team_df["team"] == team) & (team_df["date"] < date)].sort_values(by=['date'], ascending=False)
   last5 = past_games.head(5)
   goals = past_games["score"].mean()
   goals_l5 = last5["score"].mean()
   goals_suf = past_games["suf_score"].mean()
   goals_suf_l5 = last5["suf_score"].mean()
   rank = past_games["suf_rank"].mean()
   rank_l5 = last5["suf_rank"].mean()
   if len(last5) > 0:
    points = past_games["total_points"].values[0] - past_games["total_points"].values[-1]
    points_l5 = last5["total_points"].values[0] - last5["total_points"].values[-1] 
   else:
    points = 0
    points_l5 = 0
   gp = past_games["team_points"].mean()
   gp_l5 = last5["team_points"].mean()
   gp_rank = past_games["points_by_rank"].mean()
   gp_rank_l5 = last5["points_by_rank"].mean()
   
   new_val.append([goals, goals_l5, goals_suf, goals_suf_l5, rank, rank_l5, points, points_l5, gp, gp_l5, gp_rank, gp_rank_l5])
```

```python
cols =["goals_mean", "goals_mean_l5", "goals_suf_mean", "goals_suf_mean_l5", "rank_mean", "rank_mean_l5", 
       "points_mean", "points_mean_l5", "game_points_mean", "game_points_mean_l5", "game_points_rank_mean", "game_points_rank_mean_l5"]

new_val_df = pd.DataFrame(new_val, columns=cols)
```

```python
X = predict_df.drop(columns=['home_team', 'away_team', 'target'])
y = predict_df['target']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size= 0.2, stratify=y, random_state=42)
```

```python
gb = GradientBoostingClassifier(random_state=42)

params = {"learning_rate": [0.01, 0.1, 0.5],
            "min_samples_split": [5, 10],
            "min_samples_leaf": [3, 5],
            "max_depth":[3,5,10],
            "max_features":["sqrt"],
            "n_estimators":[100, 200]
         } 

gb_cv = GridSearchCV(gb, params, cv = 3, n_jobs = -1, verbose = False)

gb_cv.fit(X_train.values, np.ravel(y_train))
gb = gb_cv.best_estimator_
```

```python
params_rf = {
    "max_depth":        [5, 8, 10],       # was 20 — way too deep
    "min_samples_split":[20, 30],         # was 10 — increase to avoid small splits
    "max_leaf_nodes":   [50, 100],        # was 175 — fewer leaves
    "min_samples_leaf": [10, 20],         # was 5 — larger leaf size
    "n_estimators":     [200, 300],
    "max_features":     ["sqrt"],
    "class_weight":     ["balanced"],     # handles 131/94 imbalance
}

rf = RandomForestClassifier(random_state=42)

rf_cv = GridSearchCV(rf, params_rf, cv=3, 
                     n_jobs=-1, verbose=True,
                     scoring='roc_auc')

rf_cv.fit(X_train.values, np.ravel(y_train))
rf = rf_cv.best_estimator_

print(f"Best params: {rf_cv.best_params_}")
print(f"Best AUC:    {rf_cv.best_score_:.4f}")
```

```python

params_xgb = {
    "n_estimators":     [100, 200],
    "max_depth":        [2, 3],
    "learning_rate":    [0.01, 0.05],
    "subsample":        [0.6, 0.7],
    "colsample_bytree": [0.6, 0.7],
    "min_child_weight": [5, 10],
    "reg_alpha":        [0.5, 1.0],
    "reg_lambda":       [2.0, 5.0],
    "gamma":            [0.3, 0.5],
}

xgb = XGBClassifier(
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

xgb_cv = GridSearchCV(xgb, params_xgb, cv=3,
                       n_jobs=-1, verbose=True,
                       scoring='roc_auc')

xgb_cv.fit(X_train.values, np.ravel(y_train))
xgb = xgb_cv.best_estimator_

print(f"Best params: {xgb_cv.best_params_}")
print(f"Best AUC:    {xgb_cv.best_score_:.4f}")
```

```python
def analyze(model):
    fpr, tpr, _ = roc_curve(y_test, model.predict_proba(X_test.values)[:,1])
    fpr_train, tpr_train, _ = roc_curve(y_train, model.predict_proba(X_train.values)[:,1])
    
    auc_test = roc_auc_score(y_test, model.predict_proba(X_test.values)[:,1])
    auc_train = roc_auc_score(y_train, model.predict_proba(X_train.values)[:,1])
    
    cm = confusion_matrix(y_test, model.predict(X_test.values))

    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    
    axes[0].plot([0, 1], [0, 1], 'k--')
    axes[0].plot(fpr, tpr, label=f"Test (AUC: {auc_test:.2f})")
    axes[0].plot(fpr_train, tpr_train, label=f"Train (AUC: {auc_train:.2f})")
    axes[0].set_title('ROC Curves')
    axes[0].set_xlabel('False Positive Rate')
    axes[0].set_ylabel('True Positive Rate')
    axes[0].legend()

    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=axes[1])
    axes[1].set_title('Test Confusion Matrix')
    axes[1].set_xlabel('Predicted')
    axes[1].set_ylabel('Actual')

    plt.tight_layout()
    plt.show()
```

```python
def get_win_prob(team1, team2):
    s1 = find_stats_safe(team1)
    s2 = find_stats_safe(team2)
    if s1 is None or s2 is None:
        return 0.5, 0.5
    f1 = find_features(s1, s2)
    f2 = find_features(s2, s1)
    p1 = gb.predict_proba([f1])
    p2 = gb.predict_proba([f2])
    t1 = (p1[0][0] + p2[0][1]) / 2
    t2 = (p2[0][0] + p1[0][1]) / 2
    total = t1 + t2
    return round(t1/total, 2), round(t2/total, 2)

def simulate_round(matchups, round_name):
    games, winners = [], []
    print(f"\n{'='*10} {round_name} {'='*10}")
    for team1, team2 in matchups:
        p1, p2 = get_win_prob(team1, team2)
        winner = team1 if p1 >= p2 else team2
        winners.append(winner)
        games.append((team1, team2, (p1, p2)))
        print(f"  {team1} ({p1}) vs {team2} ({p2}) → {winner}")
    
    # ensure even number for next round
    if len(winners) % 2 != 0:
        print(f"⚠️ Odd number of winners in {round_name}: {len(winners)}")
    return games, winners
```

Apply Monte Carlo to predict the prob of winner

```python
def get_model_probabilities(team1, team2):
    """PLACEHOLDER: Replace this with your actual Gradient Boosting model.

    This function should take two team names, look up their current stats,
    and pass them to your model to get real, dynamic win probabilities.
    """

    base_probabilities = {
        "Spain": 0.51,
        "France": 0.50,
        "Brazil": 0.50
    }

    p1 = base_probabilities.get(team1, 0.50)
    p2 = base_probabilities.get(team2, 0.50)

    # Normalize to make sure they add up exactly to 1.0
    total = p1 + p2
    return (p1 / total, p2 / total)


N_SIM = 500000  # Updated to your target simulation count
champions = []

bracket = {
    "QF": [
        ("Spain", "Germany", get_model_probabilities("Spain", "Germany")),
        ("Morocco", "Portugal", get_model_probabilities("Morocco", "Portugal")),
        ("Brazil", "Argentina", get_model_probabilities("Brazil", "Argentina")),
        ("France", "England", get_model_probabilities("France", "England")),
    ]
}

# MONTE CARLO SIMULATION LOOP
for _ in range(N_SIM):

    current_round = bracket["QF"]

    while True:
        winners = []

        # Simulate the matchups in the current round
        for team1, team2, (p1, p2) in current_round:
            winner = random.choices([team1, team2], weights=[p1, p2])[0]
            winners.append(winner)

        # Check if we have crowned a single tournament champion
        if len(winners) == 1:
            champions.append(winners[0])
            break

        next_round = []
        for i in range(0, len(winners), 2):
            t1 = winners[i]
            t2 = winners[i + 1]

            prob_t1, prob_t2 = get_model_probabilities(t1, t2)

            next_round.append((t1, t2, (prob_t1, prob_t2)))

        current_round = next_round

champ_counts = Counter(champions)

champ_probs = pd.DataFrame(
    champ_counts.items(), columns=["Team", "Championship Probability"]
)

# Convert count to percentage based on total simulations
champ_probs["Championship Probability"] = (
    champ_probs["Championship Probability"] / N_SIM * 100
)

# Sort results cleanly by top probabilities
champ_probs = champ_probs.sort_values(
    "Championship Probability", ascending=False
).reset_index(drop=True)

# Format column display to look clean
champ_probs["Championship Probability"] = champ_probs[
    "Championship Probability"
].map("{:.2f}%".format)

print("\n=== WORLD CUP 2026 SIMULATION RESULTS ===")
print(champ_probs)
```

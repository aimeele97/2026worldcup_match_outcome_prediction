## 1.3 Data Preprocessing

```python
# clean and merge the ranking datasets

rank_2024_df['rank_date'] = pd.to_datetime(rank_2024_df['rank_date'])
mask = rank_2024_df['rank_date'] >= pd.to_datetime('2010-01-01')
rank_2024_df = rank_2024_df[mask].reset_index(drop=True)
```

```python
date_range = pd.date_range(rank_2024_df['rank_date'].min(), rank_2024_df['rank_date'].max(), freq='D')
countries = rank_2024_df['country_full'].unique()

full_grid = pd.MultiIndex.from_product([countries, date_range], names=['country_full', 'rank_date'])
full_grid_df = pd.DataFrame(index=full_grid).reset_index()
new_rank_df = pd.merge(full_grid_df, rank_2024_df, on=['country_full', 'rank_date'], how='left')

# forward fill per country — carries previous rank forward
new_rank_df = new_rank_df.sort_values(['country_full', 'rank_date'])
new_rank_df = new_rank_df.set_index('rank_date')
new_rank_df = new_rank_df.groupby('country_full')[new_rank_df.columns].ffill()
new_rank_df = new_rank_df.reset_index()

print(new_rank_df.columns.tolist())
new_rank_df.head()
```

```python
# add prefix to ranking columns before merge
home_rank_df = new_rank_df.add_prefix('home_')
away_rank_df = new_rank_df.add_prefix('away_')

merged_df = pd.merge(result_df, home_rank_df,
         left_on=['home_team', 'date'],
         right_on=['home_country_full', 'home_rank_date'],
         how='left')\
        .merge(away_rank_df,
         left_on=['away_team', 'date'],
         right_on=['away_country_full', 'away_rank_date'],
         how='left')
```

```python
# the ranking data after 2024-06-20 is not available, i found the other data for ranking each team before world cup 2026 
# we will use these scores and rank for the missing ranking after 2024-06-20

rank_2026_df.rename(columns={'team': 'country_full', 'team_code': 'country_abrv', 'association': 'confederation'}, inplace=True)
rank_2026_df['rank_change'] = rank_2026_df['rank'] - rank_2026_df['previous_rank']
```

```python
temp_df = merged_df[merged_df['date'] > pd.to_datetime('2024-06-20')]

# create a lookup dict from your ranking df
rank_lookup = rank_2026_df.set_index('country_full')['rank'].to_dict()
previous_rank_lookup = rank_2026_df.set_index('country_full')['previous_rank'].to_dict()
points_lookup = rank_2026_df.set_index('country_full')['points'].to_dict()
previous_points_lookup = rank_2026_df.set_index('country_full')['previous_points'].to_dict()
rank_change_lookup = rank_2026_df.set_index('country_full')['rank_change'].to_dict()
confederation_lookup = rank_2026_df.set_index('country_full')['confederation'].to_dict()

# fill NaN by mapping country name
temp_df['home_rank'] = temp_df['home_rank'].fillna(temp_df['home_team'].map(rank_lookup))
temp_df['away_rank'] = temp_df['away_rank'].fillna(temp_df['away_team'].map(rank_lookup))
temp_df['home_total_points'] = temp_df['home_total_points'].fillna(temp_df['home_team'].map(points_lookup))
temp_df['away_total_points'] = temp_df['away_total_points'].fillna(temp_df['away_team'].map(points_lookup))
temp_df['home_previous_points'] = temp_df['home_previous_points'].fillna(temp_df['home_team'].map(previous_points_lookup))
temp_df['away_previous_points'] = temp_df['away_previous_points'].fillna(temp_df['away_team'].map(previous_points_lookup))
temp_df['home_rank_change'] = temp_df['home_rank_change'].fillna(temp_df['home_team'].map(rank_change_lookup))
temp_df['away_rank_change'] = temp_df['away_rank_change'].fillna(temp_df['away_team'].map(rank_change_lookup))
temp_df['home_confederation'] = temp_df['home_confederation'].fillna(temp_df['home_team'].map(confederation_lookup))
temp_df['away_confederation'] = temp_df['away_confederation'].fillna(temp_df['away_team'].map(confederation_lookup))
```

```python
# split, modify, recombine
before = merged_df[merged_df['date'] <= pd.to_datetime('2024-06-20')]
after = temp_df 

merged_df = pd.concat([before, after]).sort_values('date').reset_index(drop=True)
merged_df.head()
```

```python
# only use the data from 2022-06-11 to 2026-06-10 for model training, as the ranking data before 2022 is not relevant to the current team performance and may introduce noise
final_df = merged_df.copy()
final_df.drop(columns=['home_country_full', 'home_rank_date', 'away_country_full', 'away_rank_date', 'home_country_abrv', 'away_country_abrv'], inplace=True)
print(final_df.columns)

# keep only 2026 world cup teams 
final_df = final_df[final_df['home_team'].isin(team_2026_list)].reset_index(drop=True)

# remove 2026 world cup matches data
world_cup_2026_df = final_df[final_df['date'] >= pd.to_datetime('2026-06-11')].reset_index(drop=True)
final_df = final_df[final_df['date'] < pd.to_datetime('2026-06-11')].reset_index(drop=True)
final_df = final_df[final_df['date'] >= pd.to_datetime('2022-06-11')].reset_index(drop=True)
```

## 1.4 Feature Engineering

```python
# Creating new features based on domain-specific knowledge

def assign_result(home_score, away_score):
    if home_score > away_score:
        return pd.Series([0, 3, 0])
    if home_score < away_score:
        return pd.Series([1, 0, 3])
    else:
        return pd.Series([2, 1, 1])

results = final_df.apply(lambda x: assign_result(x["home_score"], x["away_score"]), axis=1)
final_df[['result', 'home_team_points', 'away_team_points']] = results
```

```python
final_df["rank_dif"] = final_df["home_rank"] - final_df["away_rank"]
final_df["score_dif"] = final_df["home_score"] - final_df["away_score"]
final_df["points_home_by_rank"] = final_df["home_team_points"]/final_df["away_rank"]
final_df["points_away_by_rank"] = final_df["away_team_points"]/final_df["home_rank"]
```

```python
home_df = final_df[['date','home_team', 'home_score', 'away_score', 'home_rank', 'away_rank', 'home_total_points',
       'home_previous_points', 'home_rank_change', 'home_confederation',
       'home_team_points', 'points_home_by_rank', 'result', 'rank_dif']]
home_df.columns = [h.replace("home_", "").replace("_home", "").replace("away_", "suf_").replace("_away", "_suf") for h in home_df.columns]
away_df = final_df[['date','away_team', 'away_score', 'home_score', 'away_rank', 'home_rank', 'away_total_points',
       'away_previous_points', 'away_rank_change', 'away_confederation',
       'away_team_points', 'points_away_by_rank', 'result', 'rank_dif']]
away_df.columns = [h.replace("away_", "").replace("_away", "").replace("home_", "suf_").replace("_home", "_suf") for h in away_df.columns]
team_df = pd.concat([home_df, away_df], ignore_index=True)
```

```python
new_val = []
for idx, row in team_df.iterrows():
   team= row['team']
   date = row['date']
   past_games = team_df.loc[(team_df["team"] == team) & (team_df["date"] < date)].sort_values(by=['date'], ascending=False)
   last5 = past_games.head(5)
   goals = past_games["score"].mean()
   goals_l5 = last5["score"].mean()
   goals_suf = past_games["suf_score"].mean()
   goals_suf_l5 = last5["suf_score"].mean()
   rank = past_games["suf_rank"].mean()
   rank_l5 = last5["suf_rank"].mean()
   if len(last5) > 0:
    points = past_games["total_points"].values[0] - past_games["total_points"].values[-1]
    points_l5 = last5["total_points"].values[0] - last5["total_points"].values[-1] 
   else:
    points = 0
    points_l5 = 0
   gp = past_games["team_points"].mean()
   gp_l5 = last5["team_points"].mean()
   gp_rank = past_games["points_by_rank"].mean()
   gp_rank_l5 = last5["points_by_rank"].mean()
   
   new_val.append([goals, goals_l5, goals_suf, goals_suf_l5, rank, rank_l5, points, points_l5, gp, gp_l5, gp_rank, gp_rank_l5])
```

```python
cols =["goals_mean", "goals_mean_l5", "goals_suf_mean", "goals_suf_mean_l5", "rank_mean", "rank_mean_l5", 
       "points_mean", "points_mean_l5", "game_points_mean", "game_points_mean_l5", "game_points_rank_mean", "game_points_rank_mean_l5"]

new_val_df = pd.DataFrame(new_val, columns=cols)
```

```python
agg_team_df = pd.concat([team_df, new_val_df], axis=1).reset_index(drop=True)
agg_team_df.columns[-12:]
home_agg_df = agg_team_df[:int(len(agg_team_df)/2)]
away_agg_df = agg_team_df[int(len(agg_team_df)/2):].reset_index(drop=True)

# get only the nw cols to merge back to final df
home_agg_df = home_agg_df.iloc[:, -12:]
away_agg_df = away_agg_df.iloc[:, -12:]

home_agg_df.columns = ['home_' + str(col) for col in home_agg_df.columns]
away_agg_df.columns = ['away_' + str(col) for col in away_agg_df.columns]
```

```python
temp_agg_df = pd.concat([home_agg_df, away_agg_df], axis=1)
final_team_df = pd.concat([final_df, temp_agg_df], axis=1).reset_index(drop=True)
final_team_df['is_friendly'] = [1 if val == 'Friendly' else 0 for val in final_team_df['tournament']]
final_team_df = pd.get_dummies(final_team_df, columns=["is_friendly"])
```

# Part 2: Predictive Model Development

```python
# helper function to visualize model performance

def analyze(model):
    fpr, tpr, _ = roc_curve(y_test, model.predict_proba(X_test.values)[:,1])
    fpr_train, tpr_train, _ = roc_curve(y_train, model.predict_proba(X_train.values)[:,1])
    
    auc_test = roc_auc_score(y_test, model.predict_proba(X_test.values)[:,1])
    auc_train = roc_auc_score(y_train, model.predict_proba(X_train.values)[:,1])
    
    cm = confusion_matrix(y_test, model.predict(X_test.values))

    fig, axes = plt.subplots(1, 2, figsize=(15, 6))
    
    axes[0].plot([0, 1], [0, 1], 'k--')
    axes[0].plot(fpr, tpr, label=f"Test (AUC: {auc_test:.2f})")
    axes[0].plot(fpr_train, tpr_train, label=f"Train (AUC: {auc_train:.2f})")
    axes[0].set_title('ROC Curves')
    axes[0].set_xlabel('False Positive Rate')
    axes[0].set_ylabel('True Positive Rate')
    axes[0].legend()

    sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", ax=axes[1])
    axes[1].set_title('Test Confusion Matrix')
    axes[1].set_xlabel('Predicted')
    axes[1].set_ylabel('Actual')

    plt.tight_layout()
    plt.show()
```

## 2.1 Gradent Boosting

```python
gb = GradientBoostingClassifier(random_state=42)
params_gb = {"learning_rate": [0.01, 0.1, 0.5],
            "min_samples_split": [5, 10],
            "min_samples_leaf": [3, 5],
            "max_depth":[3,5,10],
            "max_features":["sqrt"],
            "n_estimators":[100, 200]
         } 

gb_cv = GridSearchCV(gb, params_gb, cv = 3, n_jobs = -1, verbose = False)

gb_cv.fit(X_train.values, np.ravel(y_train))
best_gb = gb_cv.best_estimator_

analyze(best_gb)
```

```python
rf = RandomForestClassifier(random_state=42)
params_rf = {
    "max_depth":        [5, 8, 10],       # was 20 — way too deep
    "min_samples_split":[20, 30],         # was 10 — increase to avoid small splits
    "max_leaf_nodes":   [50, 100],        # was 175 — fewer leaves
    "min_samples_leaf": [10, 20],         # was 5 — larger leaf size
    "n_estimators":     [200, 300],
    "max_features":     ["sqrt"],
    "class_weight":     ["balanced"],     # handles 131/94 imbalance
}

rf_cv = GridSearchCV(rf, params_rf, cv=3, 
                     n_jobs=-1, verbose=True,
                     scoring='roc_auc')

rf_cv.fit(X_train.values, np.ravel(y_train))
best_rf = rf_cv.best_estimator_

analyze(best_rf)
```

```python
# define a function to predict team winner each match
def find_stats(team_1):    
    past_games = team_df[team_df["team"] == team_1].sort_values("date")
    last5 = past_games.tail(5)

    if len(past_games) == 0:
        print(f"WARNING: No data found for '{team_1}'")
        return [100, 1.0, 1.0, 1.0, 1.0, 50, 50, 1.0, 1.0]  # neutral rank=100, not 0

    team_1_rank = past_games["rank"].values[-1]
    team_1_goals = past_games.score.mean()
    team_1_goals_l5 = last5.score.mean()
    team_1_goals_suf = past_games.suf_score.mean()
    team_1_goals_suf_l5 = last5.suf_score.mean()
    team_1_rank_suf = past_games.suf_rank.mean()
    team_1_rank_suf_l5 = last5.suf_rank.mean()
    team_1_gp_rank = past_games.points_by_rank.mean()
    team_1_gp_rank_l5 = last5.points_by_rank.mean()

    return [team_1_rank, team_1_goals, team_1_goals_l5, team_1_goals_suf, 
            team_1_goals_suf_l5, team_1_rank_suf, team_1_rank_suf_l5, team_1_gp_rank, team_1_gp_rank_l5]

def find_features(team_1, team_2):
    rank_dif = team_1[0] - team_2[0]
    goals_dif = team_1[1] - team_2[1]
    goals_dif_l5 = team_1[2] - team_2[2]
    goals_suf_dif = team_1[3] - team_2[3]
    goals_suf_dif_l5 = team_1[4] - team_2[4]
    goals_per_ranking_dif = (team_1[1]/team_1[5]) - (team_2[1]/team_2[5])
    dif_rank_agst = team_1[5] - team_2[5]
    dif_rank_agst_l5 = team_1[6] - team_2[6]
    dif_gp_rank = team_1[7] - team_2[7]
    dif_gp_rank_l5 = team_1[8] - team_2[8]
    
    return [rank_dif, goals_dif, goals_dif_l5, goals_suf_dif, goals_suf_dif_l5, goals_per_ranking_dif, 
            dif_rank_agst, dif_rank_agst_l5, dif_gp_rank, dif_gp_rank_l5, 1, 0]

def get_win_prob(team1, team2):
    s1 = find_stats(team1)
    s2 = find_stats(team2)
    if s1 is None or s2 is None:
        return 0.5, 0.5
    
    f1 = find_features(s1, s2)
    f2 = find_features(s2, s1)

    p1 = best_xgb.predict_proba([f1]) # apply XGB best params model to predict the winning team
    p2 = best_xgb.predict_proba([f2]) # apply XGB best params model to predict the winning team

    t1 = (p1[0][0] + p2[0][1]) / 2
    t2 = (p2[0][0] + p1[0][1]) / 2

    total = t1 + t2
    
    return round(t1 / total, 2), round(t2 / total, 2)
```

```python
advanced_group = []
group_results = {}
group_standings = {}

for group, teams in groups.items():

    table = {team: [0, []] for team in teams}
    group_results[group] = []

    # Play every fixture
    for team1, team2 in combinations(teams, 2):
        p1, p2 = get_win_prob(team1, team2)

        # Award points
        if abs(p1 - p2) < 0.05:
            result = "Draw"
            table[team1][0] += 1
            table[team2][0] += 1

        elif p1 > p2:
            result = team1
            table[team1][0] += 3

        else:
            result = team2
            table[team2][0] += 3

        # Save probabilities
        table[team1][1].append(p1)
        table[team2][1].append(p2)

        # Save fixture
        group_results[group].append({
            "team1": team1,
            "team2": team2,
            "team1_prob": p1,
            "team2_prob": p2,
            "winner": result
        })

    # Build final standings
    final_table = [{"team": team, "points": points} for team, (points, _) in table.items()]
    final_table.sort(key=lambda x: x["points"], reverse=True)

    # Save standings
    group_standings[group] = final_table

    # Save top 2
    advanced_group.append([final_table[0]["team"],final_table[1]["team"]])
```

```python
round_of_32 = [
    ("Match_73", second_place['A'], second_place['B']),     # 2A vs 2B
    ("Match_74", first_place['E'],  best_third_place['D']), # 1E vs 3ABCDF
    ("Match_75", first_place['F'],  second_place['C']),     # 1F vs 2C
    ("Match_76", first_place['C'],  second_place['F']),     # 1C vs 2F
    ("Match_77", first_place['I'],  best_third_place['H']), # FIXED: 1I vs 3CDFGH
    ("Match_78", second_place['E'], second_place['I']),     # 2E vs 2I
    ("Match_79", first_place['A'],  best_third_place['C']), # 1A vs 3CEFHI
    ("Match_80", first_place['L'],  best_third_place['I']), # 1L vs 3EHIJK
    ("Match_81", first_place['D'],  best_third_place['B']), # 1D vs 3BEFIJ
    ("Match_82", first_place['G'],  best_third_place['A']), # 1G vs 3AEHIJ
    ("Match_83", second_place['K'], second_place['L']),     # 2K vs 2L
    ("Match_84", first_place['H'],  second_place['J']),     # 1H vs 2J
    ("Match_85", first_place['B'],  best_third_place['G']), # 1B vs 3EFGIJ
    ("Match_86", first_place['J'],  second_place['H']),     # 1J vs 2H
    ("Match_87", first_place['K'],  best_third_place['E']), # 1K vs 3DEIJL
    ("Match_88", second_place['D'], second_place['G']),     # 2D vs 2G
]

ROUND_OF_16_STRUCTURE = [
    ("Match_89", "Match_74", "Match_77"),     
    ("Match_90", "Match_73", "Match_75"),
    ("Match_91", "Match_83", "Match_84"), 
    ("Match_92", "Match_81", "Match_82"),   
    ("Match_93", "Match_79", "Match_80"), 
    ("Match_94", "Match_76", "Match_78"),   
    ("Match_95", "Match_85", "Match_86"), 
    ("Match_96", "Match_87", "Match_88"),
]

QUARTERFINALS_STRUCTURE = [
    ("Match_97", "Match_89", "Match_90"),     
    ("Match_98", "Match_91", "Match_92"),
    ("Match_99", "Match_93", "Match_94"), 
    ("Match_100", "Match_95", "Match_96")
]

SEMIFINALS_STRUCTURE = [
    ("Match_101", "Match_97", "Match_98"),     
    ("Match_102", "Match_99", "Match_100")
]

playoffs = {"R32": [], "R16": [], "QF": [], "SF": [], "Final": []}
all_tournament_results = {}

def simulate_bracket_stage(stage_name: str, stage_structure: list, previous_winners: dict):
    """Simulates a stage of the bracket using team predictions."""
    print(f"\n==========================================")
    print(f"      {stage_name.upper()}")
    print(f"==========================================")

    current_winners = {}
    current_losers = {}
    stage_match_tuples = []

    for next_match_id, feeder1_id, feeder2_id in stage_structure:
        team1 = previous_winners[feeder1_id]
        team2 = previous_winners[feeder2_id]

        prob1, prob2 = get_win_prob(team1, team2)
        stage_match_tuples.append((team1, team2, (prob1, prob2)))

        if prob1 >= prob2:
            winner, winner_prob = team1, prob1
            loser = team2
        else:
            winner, winner_prob = team2, prob2
            loser = team1

        current_winners[next_match_id] = winner
        current_losers[f"{next_match_id}_loser"] = loser

        print(
            f"{next_match_id}: {team1} ({prob1*100:.0f}%) vs {team2} ({prob2*100:.0f}%) "
            f"--> Winner: {winner} ({winner_prob*100:.0f}%)"
        )

    return current_winners, current_losers, stage_match_tuples

# Round of 32
r32_winners = {}
for match_id, team1, team2 in round_of_32:
    prob1, prob2 = get_win_prob(team1, team2)
    playoffs["R32"].append((team1, team2, (prob1, prob2)))
    winner = team1 if prob1 >= prob2 else team2
    r32_winners[match_id] = winner

all_tournament_results.update(r32_winners)

# Round of 16
r16_winners, _, playoffs["R16"] = simulate_bracket_stage(
    "Round of 16", ROUND_OF_16_STRUCTURE, all_tournament_results
)
all_tournament_results.update(r16_winners)

# Quarterfinals
qf_winners, _, playoffs["QF"] = simulate_bracket_stage(
    "Quarterfinals", QUARTERFINALS_STRUCTURE, all_tournament_results
)
all_tournament_results.update(qf_winners)

# Semifinals
sf_winners, sf_losers, playoffs["SF"] = simulate_bracket_stage(
    "Semifinals", SEMIFINALS_STRUCTURE, all_tournament_results
)
all_tournament_results.update(sf_winners)
all_tournament_results.update(sf_losers)  # Populates 'Match_101_loser' and 'Match_102_loser'

# Third Place Play-off
match_103_team1 = all_tournament_results["Match_101_loser"]
match_103_team2 = all_tournament_results["Match_102_loser"]
p1, p2 = get_win_prob(match_103_team1, match_103_team2)
third_place_winner = match_103_team1 if p1 >= p2 else match_103_team2

print(
    f"\n3rd Place Match: {match_103_team1} ({p1*100:.0f}%) vs {match_103_team2} ({p2*100:.0f}%) "
    f"--> 3rd Place: {third_place_winner}"
)

# Final 
final_team1 = all_tournament_results["Match_101"]
final_team2 = all_tournament_results["Match_102"]
p1, p2 = get_win_prob(final_team1, final_team2)
champion = final_team1 if p1 >= p2 else final_team2
playoffs["Final"] = [(final_team1, final_team2, (p1, p2))]

print(
    f"\nWorld Cup Final (Match 104): {final_team1} ({p1*100:.0f}%) vs {final_team2} ({p2*100:.0f}%)\n"
)
print(f"🏆 World Cup 2026 Champion: {champion}")
```

# Part 4: Monte Carlo Simulation

**Utilizing Elgo Rating Formula to run Monte Carlo simulation (1,000 trials) on Model predict 8 quarter final teams**

- Definition: According to Wikipedia: " The Elo[a] rating system is a method for calculating the relative skill levels of players, originally designed for rating chess players. It is a special case of the Bradley–Terry model. The Elo system was invented as an improved chess rating system over the previously used Harkness rating system[1] and has since been adapted for use in other zero-sum games and sports, including tennis, association football (soccer), American football, baseball, basketball, pool, various board games and esports. It is named after its creator Arpad Elo, a Hungarian-American chess master and physics professor."

```python
# extract the points for 2026 world cup teams
rank_df = pd.read_csv('data/fifa_ranking_2026-06-08.csv')
rank_df['team'] = rank_df['team'].replace(name_map)
wc_team = rank_df[rank_df['team'].isin(team_2026_list)]
wc_team.reset_index(drop=True, inplace=True)
```

```python
# BASE PROBABILITY: Elo-style logistic transform of FIFA points
random.seed(42)

base_points = dict(zip(wc_team['team'], wc_team['points']))

def get_model_probabilities(team1, team2):
    p1_points = base_points[team1]
    p2_points = base_points[team2]
    p1 = 1 / (1 + 10 ** ((p2_points - p1_points) / 400))
    return (p1, 1 - p1)

# MONTE CARLO SIMULATION
N_SIM = 1000
champions = []

for _ in range(N_SIM):
    current_round = playoffs['QF']

    while True:
        winners = []
        for team1, team2, _ in current_round:
            p1, p2 = get_model_probabilities(team1, team2)
            winner = random.choices([team1, team2], weights=[p1, p2])[0]
            winners.append(winner)

        if len(winners) == 1:
            champions.append(winners[0])
            break

        next_round = []
        for i in range(0, len(winners), 2):
            t1 = winners[i]
            t2 = winners[i + 1]
            prob_t1, prob_t2 = get_model_probabilities(t1, t2)
            next_round.append((t1, t2, (prob_t1, prob_t2)))
        current_round = next_round

# RESULTS
champ_counts = Counter(champions)
champ_probs = pd.DataFrame(champ_counts.items(), columns=["Team", "Winning prob (%)"])
champ_probs["Winning prob (%)"] = (champ_probs["Winning prob (%)"] / N_SIM * 100).round(2)
champ_probs = champ_probs.sort_values("Winning prob (%)", ascending=False).reset_index(drop=True)

print("\n=== WORLD CUP 2026 SIMULATION RESULTS ===")
champ_probs
```

```python
plot_df = champ_probs.sort_values(by="Winning prob (%)")
n = len(plot_df)

colors = (["#9CA3AF"] * max(0, n - 4) + ["#3B82F6"] * min(3, max(0, n - 1)) + ["#10B981"])

ax = plot_df.set_index("Team")["Winning prob (%)"].plot(kind="barh", color=colors, figsize=(9, 5) )

for p in ax.patches:
  width = p.get_width()
  ax.annotate(f" {width:.2f}%", (width, p.get_y() + p.get_height() / 2), va="center")

plt.title("World Cup 2026 Champion Prediction (Probabilities of championship based on Mote Carlo simulation)", fontsize=12)
plt.xlim(0, max(plot_df["Winning prob (%)"]) * 1.15)
plt.show()
```
